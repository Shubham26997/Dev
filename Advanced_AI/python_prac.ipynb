```python
arr = [20,20,30,80]
total_sum = 0
for each in range(len(arr)):
    total_sum += arr[each]

total_sum
```

```python
arr = [2,3,10,20]
multi_arr = 1 # if 0 then final answer will be 0 so we kept it 1
for each in arr:
    multi_arr *= each

multi_arr
```

```python
arr = [[1,2,3,4], [6,7,8], [4,5,6,7]]

for row in range(len(arr)):
    for col in range(len(arr[row])):
        print(arr[row][col], end=' ')
    print()
```
