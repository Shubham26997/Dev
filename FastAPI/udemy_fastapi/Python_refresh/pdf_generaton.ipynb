```python
from reportlab.lib.pagesizes import A4
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, PageBreak
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.lib import colors

# Output file path
output_file = "credit_card_strategy_full.pdf"

# Document setup
doc = SimpleDocTemplate(output_file, pagesize=A4, rightMargin=30, leftMargin=30, topMargin=30, bottomMargin=18)
styles = getSampleStyleSheet()
styleN = styles["BodyText"]
styleH = styles["Heading2"]
styleB = styles["Heading3"]
styleTitle = ParagraphStyle(name='CenterTitle', parent=styles['Heading1'], alignment=TA_CENTER)

Story = []

# Title
Story.append(Paragraph("Combined Credit Card Usage Strategy (HDFC Millennia, Airtel Axis, Tata Neu Infinity)", styleTitle))
Story.append(Spacer(1, 12))

# Objective
Story.append(Paragraph("Objective:", styleH))
Story.append(Paragraph(
    "To maximize rewards, cashback, or NeuCoins across shopping categories and EMI purchases using a smart strategy for the following credit cards:",
    styleN))
cards = [
    "HDFC Millennia Credit Card",
    "Airtel Axis Bank Credit Card",
    "Tata Neu Infinity RuPay Credit Card"
]
for card in cards:
    Story.append(Paragraph(f"• {card}", styleN))
Story.append(Spacer(1, 12))

# Helper function to add card section
def add_card_section(title, data, best_for):
    Story.append(Paragraph(title, styleB))
    table = Table(data, hAlign='LEFT', colWidths=[160, 100, 230])
    table.setStyle(TableStyle([
        ('BACKGROUND', (0, 0), (-1, 0), colors.lightgrey),
        ('GRID', (0, 0), (-1, -1), 0.25, colors.grey),
        ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold')
    ]))
    Story.append(table)
    Story.append(Spacer(1, 10))
    Story.append(Paragraph("Best For:", styleB))
    for item in best_for:
        Story.append(Paragraph(f"• {item}", styleN))
    Story.append(Spacer(1, 12))

# Card Sections
hdfc_data = [
    ["Spend Type", "Cashback %", "Conditions"],
    ["Amazon, Flipkart", "5%", "Direct spend or via SmartBuy, ₹1,000 cap per cycle"],
    ["Swiggy, Zomato, Myntra", "5%", "Direct spend, included in ₹1,000 cap"],
    ["Other online/offline spends", "1%", "All non-excluded spends, ₹1,000 cap"],
    ["EMI Conversion", "No cashback", "No cashback on EMIs, even if post-facto"]
]
hdfc_best = ["Online shopping on major platforms", "Food delivery apps (Swiggy, Zomato)"]
add_card_section("1. HDFC Millennia Credit Card", hdfc_data, hdfc_best)

airtel_data = [
    ["Spend Type", "Cashback %", "Conditions"],
    ["Airtel services (recharge/bill pay)", "25%", "Up to ₹250/month; only via Airtel Thanks app"],
    ["Utility bill payments (via Airtel app)", "10%", "Up to ₹250/month; BBPS billers only"],
    ["Swiggy, Zomato, BigBasket", "10%", "Via app only; capped at ₹500/month total"],
    ["Other spends", "1%", "No cap"],
    ["EMI Conversion", "No cashback", "Cashback reversed if EMI chosen"]
]
airtel_best = ["Airtel bills & recharge", "Utility bills via Airtel app", "Swiggy/Zomato/BigBasket (within limits)"]
add_card_section("2. Airtel Axis Bank Credit Card", airtel_data, airtel_best)

tata_data = [
    ["Spend Type", "NeuCoins %", "Conditions & Limits"],
    ["Partner Tata brands (non-EMI)", "5%", "1.5% base + 3.5% additional; monthly brand cap applies"],
    ["Non-Tata brands (non-EMI)", "1.5%", "Base NeuCoins"],
    ["Merchant EMI (any brand)", "1.5%", "Only principal earns NeuCoins"],
    ["UPI via Tata Neu UPI ID", "1.5%", "Max 500 NeuCoins/month (0.5% UPI + 1% Tata Neu UPI)"],
    ["EMI via Smart/Dial an EMI", "No NeuCoins", "Excluded"]
]
tata_best = ["Shopping at BigBasket, Croma, Westside, etc.", "UPI spends with Tata Neu UPI ID"]
add_card_section("3. Tata Neu Infinity Card (RuPay)", tata_data, tata_best)

# Story.append(PageBreak())

# Category-wise strategy
Story.append(Paragraph("Category-Wise Optimal Usage Strategy", styleH))
cat_data = [
    ["Category", "Best Card to Use", "Reason"],
    ["Amazon/Flipkart", "HDFC Millennia (5%)", "Reliable direct 5% cashback"],
    ["Swiggy/Zomato", "Airtel Axis (10%)", "Higher cashback than Millennia, ₹500 monthly cap"],
    ["", "HDFC Millennia (5%)", "Use after Axis cap is exhausted"],
    ["BigBasket", "Airtel Axis (10%)", "Within ₹500 cap (combined with Swiggy/Zomato)"],
    ["Tata Neu brands", "Tata Neu Infinity (5%)", "Best structured rewards with fair usage caps"],
    ["Bill Payments", "Airtel Axis (25% / 10%)", "₹250/month cap for each (Airtel & utilities)"],
    ["UPI Transactions", "Tata Neu Infinity (1.5%)", "Use Tata Neu UPI ID for full benefit"],
    ["EMI Purchases", "Tata Neu Infinity (1.5%)", "Use only for merchant EMI, avoid post-facto EMI"]
]
cat_table = Table(cat_data, hAlign='LEFT', colWidths=[120, 150, 220])
cat_table.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.lightgrey),
    ('GRID', (0, 0), (-1, -1), 0.25, colors.grey),
    ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold')
]))
Story.append(cat_table)
# Story.append(PageBreak())

# Monthly usage flow
Story.append(Paragraph("Monthly Usage Flow Example", styleH))
steps = [
    "Pay Airtel bills via Airtel Axis → 25% cashback up to ₹250",
    "Pay electricity/gas bills via Airtel Axis → 10% cashback up to ₹250",
    "Use Airtel Axis for Swiggy/Zomato/BigBasket → 10% cashback up to ₹500",
    "Use HDFC Millennia for Amazon/Flipkart after Axis cap → 5% cashback",
    "Use Tata Neu for Tata brands → 5% NeuCoins (per brand cap applies)",
    "Use Tata Neu (RuPay) for UPI → 1.5% NeuCoins up to 500/month",
    "Avoid EMI unless necessary → Use Tata Neu for merchant EMI if needed"
]
for step in steps:
    Story.append(Paragraph(f"• {step}", styleN))
# Story.append(PageBreak())

# Notes on EMI
Story.append(Paragraph("Notes on EMI & Exceptions", styleH))
emi_notes = [
    "No card offers rewards on post-purchase EMI conversion",
    "Tata Neu offers 1.5% NeuCoins on principal for merchant EMI only",
    "All cards exclude fuel, wallet loads, rent, insurance, government, and prepaid/gift card transactions from rewards"
]
for note in emi_notes:
    Story.append(Paragraph(f"• {note}", styleN))
Story.append(Spacer(1, 12))

# Final tip
Story.append(Paragraph("Final Tip:", styleH))
tips = [
    "Use Airtel Axis for all Airtel and utility bill payments and food/grocery delivery",
    "Use HDFC Millennia for direct Amazon/Flipkart/e-commerce after Axis limits",
    "Use Tata Neu Infinity for Tata ecosystem, UPI, and EMI-based purchases"
]
for tip in tips:
    Story.append(Paragraph(f"• {tip}", styleN))

Story.append(Spacer(1, 12))
Story.append(Paragraph("This ensures:", styleN))
ensures = [
    "You hit all cashback caps",
    "Optimize NeuCoins redemption value (₹1/coin)",
    "Separate spends across platforms without overlap"
]
for e in ensures:
    Story.append(Paragraph(f"• {e}", styleN))

# Generate the PDF
doc.build(Story)
print(f"PDF generated: {output_file}")
```
